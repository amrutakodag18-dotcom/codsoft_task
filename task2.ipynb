```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load your cleaned dataset from Task 1
df = pd.read_csv("cleaned_dataset.csv")

# Display the first 5 rows to verify it loaded correctly
print("First 5 rows of your dataset:")
display(df.head())
```

```python
df.info()

display(df.describe())

display(df.describe(include=["object", "category"]))
```

```python
numerical_col = df.select_dtypes(include=[np.number]).columns[0]

plt.figure(figsize=(8, 4))
sns.histplot(df[numerical_col], kde=True, color="teal")
plt.title(f"Distribution of {numerical_col}")
plt.xlabel(numerical_col)
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df[numerical_col], color="orange")
plt.title(f"Boxplot for Outlier Detection in {numerical_col}")
plt.show()
```
