```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your dataset
df = pd.read_csv("cleaned_dataset.csv")
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Automatically find a text column and a numeric column to prevent errors
cat_col = df.select_dtypes(include=['object', 'category']).columns[0]
num_col = df.select_dtypes(include=['number']).columns[0]

# Prepare data safely
data_to_plot = df.groupby(cat_col)[num_col].mean().reset_index().head(10)

# 1. Bar Chart
plt.figure(figsize=(10, 5))
sns.barplot(x=cat_col, y=num_col, data=data_to_plot, palette='viridis')
plt.title(f'Average {num_col} by {cat_col}', fontsize=14, fontweight='bold')
plt.xlabel(cat_col, fontsize=12)
plt.ylabel(num_col, fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 📊 Insight: Bar Chart Analysis
* **Observation:** The bar chart compares the average values across different categories, highlighting which category has the highest and lowest metrics.
* **Key Takeaway:** Categories with the highest bars represent the primary drivers in the dataset, helping identify key focus areas for analysis.
*

```python
plt.figure(figsize=(10, 5))
plt.plot(df[cat_col].head(15), df[num_col].head(15), marker='o', color='b', linestyle='-')
plt.title(f'Trend of {num_col} across {cat_col}', fontsize=14, fontweight='bold')
plt.xlabel(cat_col, fontsize=12)
plt.ylabel(num_col, fontsize=12)
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### 📈 Insight: Line Chart Analysis
* **Observation:** The line chart tracks how the values fluctuate across the sequence or time period, showing distinct peaks and drops.
* **Key Takeaway:** The upward trends indicate periods of growth or higher activity, while downward slopes highlight areas that may need attention or further investigation.
*

```python
plt.figure(figsize=(6, 6))
df[cat_col].value_counts().head(5).plot.pie(
    autopct='%1.1f%%',
    colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0']
)
plt.title(f'Distribution of {cat_col}', fontsize=14, fontweight='bold')
plt.ylabel('')
plt.show()
```

### 🥧 Insight: Pie Chart Analysis
* **Observation:** The pie chart displays the proportional breakdown and percentage share of each category within the total dataset.
* **Key Takeaway:** The largest slices highlight the dominant segments, providing a clear visual representation of categorical distribution at a glance.
*

```python
plt.figure(figsize=(8, 5))
sns.histplot(df[num_col], kde=True, color='purple', bins=20)
plt.title(f'Distribution of {num_col}', fontsize=14, fontweight='bold')
plt.xlabel(num_col, fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

### 📊 Insight: Histogram Analysis
* **Observation:** The histogram illustrates the frequency distribution of the numerical variable, showing how data values are spread across different bins.
* **Key Takeaway:** The shape of the distribution helps identify where most data is concentrated (central tendency), as well as any skewness or outliers in the dataset.
*

```python
num_cols = df.select_dtypes(include=['number']).columns
if len(num_cols) >= 2:
    x_col, y_col = num_cols[0], num_cols[1]
    plt.figure(figsize=(8, 5))
    sns.scatterplot(x=x_col, y=y_col, data=df, color='teal')
    plt.title(f'{y_col} vs {x_col}', fontsize=14, fontweight='bold')
    plt.xlabel(x_col, fontsize=12)
    plt.ylabel(y_col, fontsize=12)
    plt.show()
```

### 📉 Insight: Scatter Plot Analysis
* **Observation:** The scatter plot displays the relationship between two continuous variables, showing how changes in one variable correspond to the other.
* **Key Takeaway:** The pattern of points reveals potential correlations (positive, negative, or none) and helps identify any distinct data clusters or outliers.
